Given an integer array nums, return true if any value appears more than once in the array, otherwise return false.

Example 1:

Input: nums = [1, 2, 3, 3]

Output: true

Example 2:

Input: nums = [1, 2, 3, 4]

Output: false
Constraints:

0 <= nums.length <= 10^5
-10^9 <= nums[i] <= 10^9

```python
class Solution:
    def hasDuplicate(self, nums: List[int]) -> bool:
        class occurrence_ctr():
            def __init__(self, val, occ):
                self.val = val
                self.occ = occ
        list_of_ctrs = []
        for i in range(len(nums)):
            tmp = nums[i]
            cntr = 0
            for j in range(len(nums)):
                if tmp == nums[j]:
                    cntr += 1
            print(cntr)
            list_of_ctrs.append(occurrence_ctr(tmp, cntr))
        i = 0
        dup_found = 0
        for i in range(len(list_of_ctrs)):
            if list_of_ctrs[i].occ > 1:
                dup_found = 1
        if dup_found == 1:
            return True
        else:
            return False
```

```python
class Solution:
    def hasDuplicate(self, nums: List[int]) -> bool:
        seen = set()
        for x in nums:
            if x in seen:
                return True
            seen.add(x)
        return False
```

This creates an empty set in Python. A set stores unique values, meaning it does not keep duplicates.

Let's break it down:
- seen — the variable name.
- set() — creates an empty set.
- = — assigns the empty set to the variable.
